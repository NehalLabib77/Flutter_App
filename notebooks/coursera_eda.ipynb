```python
data = data[['Course Name','Difficulty Level','Course URL','Course Description','Skills']]
```

```python
data.head()
```

```python
data = data.copy()

data["Course Name"] = (
    data["Course Name"]
    .str.replace(" ", "", regex=False)
    .str.replace(":", "", regex=False)
    .str.replace(r",{2,}", ",", regex=True)
)

data["Course Description"] = (
    data["Course Description"]
    .str.replace(r"[ :_()]", "", regex=True)
    .str.replace(r",{2,}", ",", regex=True)
)

data["Skills"] = (
    data["Skills"]
    .str.replace(r"[()]", "", regex=True)
)
```

```python
data.head()
```

```python
data['tags'] = data['Course Name'] + data['Difficulty Level'] + data['Course Description'] + data['Skills']
```

```python
data.head()
```

```python
data['tags'].iloc[0] # value at integer 0
```

```python
new_df = data[["Course Name", "Course URL", "tags"]].copy()

new_df["Course Name"] = (
    new_df["Course Name"]
    .str.replace(",", " ", regex=False)
)

new_df.rename(
    columns={
        "Course Name": "course_name",
        "Course URL": "course_url"
    },
    inplace=True  # Apply the change directly to the existing DataFrame
)
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x:x.lower())
```

```python
new_df.head()
```

```python
new_df.shape
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features = 5000, stop_words='english')
```

```python
vectors = cv.fit_transform(new_df['tags']).toarray()
```

```python
vectors[0]
```

```python
import nltk
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))

    return " ".join(y)  #stem
```

```python
new_df['tags'] = new_df['tags'].apply(stem)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
def recommend(course):
    course_index = new_df[new_df['course_name'] == course].index[0]
    distance = similarity[course_index]
    course_list = sorted(list(enumerate(distance)), reverse = True, key = lambda x:x[1])[1:7]
    
    recommended_courses = []
    for i in course_list:
        course_name = new_df.iloc[i[0]].course_name
        course_url = new_df.iloc[i[0]].course_url
        recommended_courses.append({'name': course_name, 'url': course_url})
    return recommended_courses   
```

```python
new_df.head()
```

```python
recommend("WriteAFeatureLengthScreenplayForFilmOrTelevision")
```
